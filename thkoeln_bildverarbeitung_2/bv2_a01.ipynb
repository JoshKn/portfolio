```python
import numpy as np
from mpl_toolkits import mplot3d
import matplotlib.pyplot as plt
import pandas as pd
```

## Setting up Inputs, Weights, Labels and Stepsize

```python
eta = 0.1
epochs = 100
```

## Training

```python
#np.random.seed(0) # für Reproduzierbarkeit
x1 = np.random.uniform(-5, 5, 10000)
x2 = np.random.uniform(-5, 5, 10000)

y = x1 + x2

w1 = np.random.uniform(-10, 10)
w2 = np.random.uniform(-10, 10)

w1_hist = [w1]
w2_hist = [w2]
loss_hist = []

for epoch in range(epochs):
    #shuffle inputs
    np.random.shuffle(x1)
    np.random.shuffle(x2)
    y = x1 + x2 # calculate label from shuffled inputs

    y_hat = x1 * w1 + x2 * w2   

    loss = 0.5 * np.mean((y - y_hat)**2)
    loss_hist.append(loss)

    grad_w1 = np.mean(x1 * (y_hat - y))
    grad_w2 = np.mean(x2 * (y_hat - y))

    w1 = w1 - eta * grad_w1
    w2 = w2 - eta * grad_w2

    w1_hist.append(w1)
    w2_hist.append(w2)

```

### Plotting 3D

```python
print(f"Eta: {eta}, Epochs: {epochs}")
print(f"Last Loss: {loss_hist[-1]}")

fig = plt.figure()
ax = plt.axes(projection="3d")

ax.plot3D(w1_hist[1:], w2_hist[1:], loss_hist, marker="x")
#plt.plot(w2_hist[1:], loss_hist)
plt.xlabel("w1")
plt.ylabel("w2")
plt.clabel("Loss")
plt.title("Loss Function for w1, w2")
plt.show()
```
